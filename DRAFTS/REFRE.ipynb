```python
import os

def process_directory(data_dir, output_dir):
    """
    بيعالج كل ملفات CSV في مجلد
    """
    # التأكد من وجود مجلد الإخراج
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)    # إنشاء المجلد لو مش موجود
    
    # المرور على كل الملفات
    for filename in os.listdir(data_dir):
        if filename.endswith(".csv"):
            input_path = os.path.join(data_dir, filename)
            output_path = os.path.join(output_dir, f"clean_{filename}")
            
            # قراءة ومعالجة
            with open(input_path, "r") as infile:
                with open(output_path, "w") as outfile:
                    for line in infile:
                        # تنظيف: إزالة المسافات الزائدة
                        clean_line = line.strip() + "\n"
                        outfile.write(clean_line)
            
            print(f"✅ تمت معالجة: {filename}")

process_directory("/data/raw/", "/data/clean/")
```

```python
# ========== استيراد Modules ==========
import os               # للتعامل مع نظام الملفات
import json             # للتعامل مع JSON
from datetime import datetime   # للتاريخ والوقت
import random           # للأرقام العشوائية (مفيد في الـ sampling)

# alias
import pandas as pd
import numpy as np

# ========== OOP - بناء Classes للـ ETL ==========
class DataPipeline:
    """
    Class بتمثل pipeline لمعالجة البيانات
    """
    def __init__(self, name, source, target):
        # Constructor - بيتنفذ لما بننشئ object
        self.name = name
        self.source = source
        self.target = target
        self.records_processed = 0
        self.start_time = None
    
    def run(self):
        # method بتشغل الـ pipeline
        self.start_time = datetime.now()
        print(f"🚀 Starting pipeline: {self.name}")
        print(f"   Source: {self.source}")
        print(f"   Target: {self.target}")
    
    def log_metrics(self, record_count):
        # method بتسجل المقاييس
        self.records_processed += record_count
        duration = (datetime.now() - self.start_time).seconds
        print(f"📊 Processed {record_count} records in {duration}s")
    
    def get_summary(self):
        # method بترجع ملخص
        return {
            "pipeline": self.name,
            "total_records": self.records_processed,
            "source": self.source,
            "target": self.target
        }

# إنشاء object من الـ Class
pipeline = DataPipeline("daily_sales", "postgres", "snowflake")
pipeline.run()
pipeline.log_metrics(50000)
print(pipeline.get_summary())

# ========== Inheritance (الوراثة) ==========
class IncrementalPipeline(DataPipeline):
    """
    Pipeline بتعمل incremental load (بتحمل التغييرات بس)
    """
    def __init__(self, name, source, target, last_run_date):
        super().__init__(name, source, target)   # استدعاء constructor الأب
        self.last_run_date = last_run_date
    
    def run(self):
        # Override - تعديل السلوك
        print(f"⏱️ Incremental load since: {self.last_run_date}")
        super().run()    # استدعاء run() من الأب

incremental = IncrementalPipeline("hourly_events", "kafka", "s3", "2024-01-01")
incremental.run()
```

```python
# ========== Pivot Table ==========
# زي Excel Pivot Table - مفيد جداً في التقارير

df = pd.DataFrame({
    "date": ["Jan", "Jan", "Feb", "Feb", "Jan", "Feb"],
    "region": ["North", "South", "North", "South", "East", "East"],
    "product": ["A", "A", "B", "B", "A", "B"],
    "sales": [100, 150, 200, 120, 80, 90]
})

# Pivot بسيط
pivot = df.pivot_table(
    values="sales",           # القيم
    index="region",           # الصفوف
    columns="date",           # الأعمدة
    aggfunc="sum"             # دالة التجميع
)
print(pivot)
# date      Feb  Jan
# region            
# East      90.0   80.0
# North    200.0  100.0
# South    120.0  150.0

# Pivot متقدم
pivot2 = df.pivot_table(
    values="sales",
    index=["region", "product"],
    columns="date",
    aggfunc=["sum", "mean"],
    fill_value=0,             # املأ الفراغات بـ 0
    margins=True              إضافة صف وعمود "All" (المجموع الكلي)
)
print(pivot2)


# ========== Timestamps ==========
# إنشاء تواريخ
ts = pd.Timestamp("2024-01-15 14:30:00")
print(ts.year)      # 2024
print(ts.month)     # 1
print(ts.day)       # 15
print(ts.dayofweek) # 0 (Monday)

# ========== Date Range ==========
# سيناريو: إنشاء جدول أيام العمل
business_days = pd.date_range(
    start="2024-01-01",
    end="2024-01-31",
    freq="B"           # B = Business Days (أيام العمل)
)
print(business_days)
# DatetimeIndex(['2024-01-01', '2024-01-02', ..., '2024-01-31'], dtype='datetime64[ns]', freq='B')

# Frequencies شائعة:
# 'D' = يومي, 'W' = أسبوعي, 'M' = شهري, 'H' = ساعة, 'T' = دقيقة

# ========== Period ==========
# فترة زمنية (شهر، ربع سنة، سنة)
period = pd.Period("2024-01", freq="M")   # يناير 2024 ككل
print(period.start_time)   # 2024-01-01
print(period.end_time)     # 2024-01-31 23:59:59

# ========== Timedelta ==========
# الفرق بين تواريخ
delta = pd.Timedelta(days=7, hours=3)
future = pd.Timestamp("2024-01-01") + delta   # 2024-01-08 03:00:00

# ========== Time Series DataFrame ==========
# سيناريو: بيانات سعر الأسهم
dates = pd.date_range("2024-01-01", periods=10, freq="D")
stock_data = pd.DataFrame({
    "date": dates,
    "open": [100, 102, 101, 105, 103, 107, 106, 108, 110, 109],
    "close": [102, 101, 105, 103, 107, 106, 108, 110, 109, 111],
    "volume": [1000, 1500, 1200, 2000, 1800, 2200, 1900, 2500, 2100, 2300]
})
stock_data.set_index("date", inplace=True)

# Indexing بالتاريخ
print(stock_data.loc["2024-01-05"])           # بيانات يوم محدد
print(stock_data.loc["2024-01-01":"2024-01-05"])  # range

# Resampling (تجميع البيانات على فترات مختلفة)
# سيناريو: بيانات يومية → شهريّة
monthly = stock_data.resample("M").agg({
    "open": "first",      # أول سعر في الشهر
    "close": "last",      # آخر سعر في الشهر
    "volume": "sum"       # إجمالي الحجم
})
print(monthly)

# Rolling Windows (المتوسط المتحرك)
stock_data["ma_3"] = stock_data["close"].rolling(window=3).mean()
# المتوسط المتحرك لآخر 3 أيام

# Shift (لحساب التغيير اليومي)
stock_data["prev_close"] = stock_data["close"].shift(1)
stock_data["daily_change"] = stock_data["close"] - stock_data["prev_close"]

def process_sensor_data():
    """
    معالجة بيانات حساسات IoT
    """
    # بيانات الحساسات (كل دقيقة)
    timestamps = pd.date_range("2024-01-01", periods=1440, freq="T")  # يوم كامل
    np.random.seed(42)
    temperatures = np.random.normal(25, 3, 1440)  # متوسط 25، انحراف 3
    
    df = pd.DataFrame({
        "timestamp": timestamps,
        "temperature": temperatures,
        "sensor_id": "SENSOR_001"
    })
    
    # 1. تعيين الـ timestamp كـ index
    df.set_index("timestamp", inplace=True)
    
    # 2. Resample لكل ساعة (متوسط الحرارة في الساعة)
    hourly = df.resample("H").agg({
        "temperature": ["mean", "min", "max", "std"]
    })
    hourly.columns = ["temp_mean", "temp_min", "temp_max", "temp_std"]
    
    # 3. اكتشاف الشذوذ (Outliers) - الحرارة اللي خارج ±2 انحراف معياري
    mean_temp = df["temperature"].mean()
    std_temp = df["temperature"].std()
    
    df["is_anomaly"] = (
        (df["temperature"] > mean_temp + 2 * std_temp) |
        (df["temperature"] < mean_temp - 2 * std_temp)
    )
    
    anomalies = df[df["is_anomaly"]]
    print(f"⚠️ تم اكتشاف {len(anomalies)} قيمة شاذة")
    
    # 4. Forward fill للقيم الناقصة
    df["temperature"] = df["temperature"].fillna(method="ffill")
    
    return df, hourly

sensor_df, hourly_df = process_sensor_data()
print(hourly_df.head())
```

```python
import pandas as pd
import numpy as np

class DataQualityEngine:
    """
    Framework كامل لفحص جودة البيانات
    """
    def __init__(self, df):
        self.df = df
        self.issues = []
    
    def check_completeness(self, threshold=0.95):
        """
        اكتمال البيانات: نسبة الـ non-null
        """
        completeness = (self.df.count() / len(self.df))
        for col, ratio in completeness.items():
            if ratio < threshold:
                self.issues.append({
                    "check": "completeness",
                    "column": col,
                    "severity": "HIGH" if ratio < 0.80 else "MEDIUM",
                    "details": f"Null ratio: {(1-ratio)*100:.1f}%"
                })
        return completeness
    
    def check_uniqueness(self, columns):
        """
        فحص التكرار في primary keys
        """
        duplicates = self.df.duplicated(subset=columns).sum()
        if duplicates > 0:
            self.issues.append({
                "check": "uniqueness",
                "columns": columns,
                "severity": "HIGH",
                "details": f"{duplicates} duplicate records found"
            })
        return duplicates
    
    def check_consistency(self, column, valid_values):
        """
        فحص التناسق: القيم المسموح بها
        """
        invalid = self.df[~self.df[column].isin(valid_values)]
        if len(invalid) > 0:
            self.issues.append({
                "check": "consistency",
                "column": column,
                "severity": "MEDIUM",
                "details": f"{len(invalid)} invalid values"
            })
        return len(invalid)
    
    def check_range(self, column, min_val, max_val):
        """
        فحص النطاق
        """
        out_of_range = self.df[
            (self.df[column] < min_val) | (self.df[column] > max_val)
        ]
        if len(out_of_range) > 0:
            self.issues.append({
                "check": "range",
                "column": column,
                "severity": "HIGH",
                "details": f"{len(out_of_range)} values out of range"
            })
        return len(out_of_range)
    
    def generate_report(self):
        """
        إنشاء تقرير
        """
        report = pd.DataFrame(self.issues)
        if len(report) == 0:
            return "✅ All checks passed!"
        return report

# استخدام
df = pd.DataFrame({
    "customer_id": [1, 2, 2, 4, 5],
    "name": ["A", "B", "C", None, "E"],
    "age": [25, 150, 30, 40, -5],
    "status": ["active", "active", "unknown", "inactive", "active"]
})

dq = DataQualityEngine(df)
dq.check_completeness(threshold=0.95)
dq.check_uniqueness(["customer_id"])
dq.check_consistency("status", ["active", "inactive"])
dq.check_range("age", 0, 120)

print(dq.generate_report())
```

```python
import pandas as pd
from datetime import datetime, timedelta

class IncrementalLoader:
    """
    تحميل تدريجي للبيانات (Delta Load)
    """
    def __init__(self, watermark_file="watermark.txt"):
        self.watermark_file = watermark_file
        self.last_load = self._read_watermark()
    
    def _read_watermark(self):
        """
        قراءة آخر timestamp تم تحميله
        """
        try:
            with open(self.watermark_file, "r") as f:
                return pd.Timestamp(f.read().strip())
        except FileNotFoundError:
            # أول مرة - نرجع تاريخ قديم
            return pd.Timestamp("1900-01-01")
    
    def _write_watermark(self, new_watermark):
        """
        تحديث الـ watermark
        """
        with open(self.watermark_file, "w") as f:
            f.write(str(new_watermark))
    
    def extract_delta(self, source_df, timestamp_column):
        """
        استخراج السجلات الجديدة بس
        """
        new_records = source_df[
            source_df[timestamp_column] > self.last_load
        ]
        print(f"📥 Found {len(new_records)} new records since {self.last_load}")
        return new_records
    
    def load(self, new_data, target_path, timestamp_column):
        """
        تحميل البيانات وتحديث الـ watermark
        """
        if len(new_data) > 0:
            # Append للبيانات القديمة
            try:
                existing = pd.read_csv(target_path)
                combined = pd.concat([existing, new_data], ignore_index=True)
            except FileNotFoundError:
                combined = new_data
            
            combined.to_csv(target_path, index=False)
            
            # تحديث الـ watermark لآخر timestamp
            new_watermark = new_data[timestamp_column].max()
            self._write_watermark(new_watermark)
            print(f"✅ Loaded {len(new_data)} records. New watermark: {new_watermark}")
        else:
            print("ℹ️ No new data to load")

# محاكاة
source = pd.DataFrame({
    "id": [1, 2, 3, 4, 5],
    "value": [100, 200, 300, 400, 500],
    "created_at": pd.to_datetime([
        "2024-01-01", "2024-01-02", "2024-01-03",
        "2024-01-04", "2024-01-05"
    ])
})

loader = IncrementalLoader()
delta = loader.extract_delta(source, "created_at")
loader.load(delta, "target.csv", "created_at")
```

```python
import time
import pandas as pd
from datetime import datetime

class PipelineMonitor:
    """
    مراقبة وقياس أداء الـ Data Pipelines
    """
    def __init__(self, pipeline_name):
        self.pipeline_name = pipeline_name
        self.metrics = []
        self.start_time = None
    
    def start(self):
        self.start_time = time.time()
        print(f"🚀 Pipeline '{self.pipeline_name}' started at {datetime.now()}")
    
    def log_stage(self, stage_name, records_in, records_out):
        """
        تسجيل مقاييس كل مرحلة
        """
        duration = time.time() - self.start_time
        rejection_rate = (records_in - records_out) / records_in if records_in > 0 else 0
        
        self.metrics.append({
            "pipeline": self.pipeline_name,
            "stage": stage_name,
            "timestamp": datetime.now(),
            "records_in": records_in,
            "records_out": records_out,
            "rejected": records_in - records_out,
            "rejection_rate": rejection_rate,
            "duration_sec": duration
        })
        
        print(f"  📊 {stage_name}: {records_in} → {records_out} "
              f"(rejected: {records_in - records_out}, rate: {rejection_rate:.2%})")
    
    def get_summary(self):
        df = pd.DataFrame(self.metrics)
        summary = {
            "total_stages": len(df),
            "total_duration": df["duration_sec"].max(),
            "total_processed": df["records_out"].sum(),
            "total_rejected": df["rejected"].sum(),
            "overall_rejection_rate": df["rejected"].sum() / df["records_in"].sum()
        }
        return summary

# استخدام
monitor = PipelineMonitor("daily_etl")
monitor.start()

# Extract
raw_data = pd.read_csv("raw.csv")
monitor.log_stage("extract", len(raw_data), len(raw_data))

# Transform
clean_data = raw_data.dropna()
monitor.log_stage("transform", len(raw_data), len(clean_data))

# Load
clean_data.to_csv("clean.csv", index=False)
monitor.log_stage("load", len(clean_data), len(clean_data))

print(monitor.get_summary())
```
